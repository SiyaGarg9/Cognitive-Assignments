```python
#Q2
import random
random.seed(1024170214)
```

```python
nums=[random.randint(100,900)for _ in range(100)]
print(nums)
```

```python
odd=[i for i in nums if i%2!=0]
print(odd)
print(len(odd))
```

```python
even=[i for i in nums if i%2==0]
print(even)
print(len(even))
```

```python
def p(n):
    if n<2:return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:return False
    return True

primes=[i for i in nums if p(i)]
print(primes)
print(len(primes))
```

```python
most=max(set(nums),key=nums.count)
print(most,nums.count(most))
```

```python
#Q3
roll="1024170214"
digits=[int(d) for d in roll[:8]]

A={d*7 for d in digits}
B={d*9 for d in digits}

print(A)
print(B)

print(A|B)
```

```python
print(A&B)
```

```python
print(A.difference(B))
print(B.difference(A))
#A.difference(B) gives elements only in A and not in B, while symmetric_difference() gives elements in either set but not in both
```

```python
print(A.symmetric_difference(B))
```

```python
print(A.issubset(B))
print(B.issuperset(A))
```
