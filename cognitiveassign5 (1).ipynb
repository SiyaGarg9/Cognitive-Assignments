```python
import numpy as np

arr=np.array([1,2,3,6,4,5])
rev=arr[::-1]
print(rev)

x=np.array([1,2,3,4,5,1,2,1,1,1])
vals,counts=np.unique(x,return_counts=True)
mf=vals[np.argmax(counts)]
idx=np.where(x==mf)
print(mf,idx)

y=np.array([1,1,1,2,3,4,2,4,3,3])
vals,counts=np.unique(y,return_counts=True)
mf=vals[np.argmax(counts)]
idx=np.where(y==mf)
print(mf,idx)
```

```python
import numpy as np

ucs420_Siya = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:",ucs420_Siya)

print("Mean:", np.mean(ucs420_Siya))
print("Median:", np.median(ucs420_Siya))
print("Maximum:", np.max(ucs420_Siya))
print("Minimum:", np.min(ucs420_Siya))
print("Unique elements:", np.unique(ucs420_Siya))

# Reshape to 4 rows and 3 columns
reshaped_ucs420_Siya = ucs420_Siya.reshape(4, 3)
print("Reshaped Array:",reshaped_ucs420_Siya)

# Resize to 2 rows and 3 columns
resized_ucs420_Siya = np.resize(ucs420_Siya, (2, 3))
print("Resized Array:",resized_ucs420_Siya)
```
