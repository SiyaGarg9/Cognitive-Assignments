```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

text = """AlphaFold is an artificial intelligence system developed by DeepMind
that predicts the three-dimensional structures of proteins. Proteins perform
many essential functions in living organisms, and their shapes influence how
they work. AlphaFold helps scientists understand protein structures much faster
than traditional experimental methods in many cases. This technology supports
research in drug discovery, disease understanding, and biological science.
I find AlphaFold fascinating because it shows how artificial intelligence can
solve complex scientific problems and contribute to human health."""

nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
# Convert text to lowercase
text = text.lower()

# Extract words containing only alphabets
words = re.findall(r'\b[a-zA-Z]+\b', text)

print("Words containing only alphabets:")
print(words)

print("\nTotal words extracted:", len(words))
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word not in stop_words
]

print("Words after removing stopwords:")
print(filtered_words)

print("\nTotal words after stopword removal:", len(filtered_words))
```

```python
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("Stemmed words:")
print(stemmed_words)
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in filtered_words
]

print("Lemmatized words:")
print(lemmatized_words)
```

```python
print(f"{'Original Word':<20} {'Stemmed Word':<20} {'Lemmatized Word':<20}")
print("-" * 60)

for original, stemmed, lemmatized in zip(
    filtered_words, stemmed_words, lemmatized_words
):
    print(f"{original:<20} {stemmed:<20} {lemmatized:<20}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd

# Three short texts of our own
texts = [
    "AlphaFold uses artificial intelligence to predict protein structures.",
    "Artificial intelligence helps doctors detect diseases and discover medicines.",
    "Protein research using AlphaFold supports scientific discovery and medicine."
]

for i, text in enumerate(texts, start=1):
    print(f"Text {i}: {text}")
```

```python
# Create CountVectorizer object
count_vectorizer = CountVectorizer()

# Generate Bag of Words representation
bow_matrix = count_vectorizer.fit_transform(texts)

# Convert to DataFrame
bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=count_vectorizer.get_feature_names_out(),
    index=["Text 1", "Text 2", "Text 3"]
)

print("Bag of Words Representation:")
print(bow_df)
```

```python
# Create TfidfVectorizer object
tfidf_vectorizer = TfidfVectorizer()

# Calculate TF-IDF scores
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

# Convert to DataFrame
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out(),
    index=["Text 1", "Text 2", "Text 3"]
)

print("TF-IDF Representation:")
print(tfidf_df.round(3))
```

```python
feature_names = tfidf_vectorizer.get_feature_names_out()

for i, row in enumerate(tfidf_matrix.toarray(), start=1):
    top_indices = row.argsort()[-3:][::-1]

    print(f"\nTop 3 keywords in Text {i}:")
    for index in top_indices:
        print(f"{feature_names[index]}: {row[index]:.3f}")
```

```python
import re
import nltk
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
# Text 1: Artificial Intelligence
text1 = """Artificial intelligence enables machines to learn from data.
AI helps computers recognize patterns and solve complex problems.
It is widely used in healthcare, education, and robotics.
AI systems can automate tasks and support decision making."""

# Text 2: Blockchain
text2 = """Blockchain is a distributed digital ledger technology.
It stores transactions securely across multiple computers.
Blockchain provides transparency and prevents data tampering.
It is widely used in cryptocurrency, finance, and smart contracts."""

print("Text 1 (AI):\n", text1)
print("\nText 2 (Blockchain):\n", text2)
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

def preprocess(text):
    # Convert to lowercase
    text = text.lower()

    # Remove punctuation and non-alphabetic characters
    text = re.sub(r'[^a-z\s]', '', text)

    # Tokenize the text
    tokens = word_tokenize(text)

    # Remove stopwords
    tokens = [word for word in tokens if word not in stop_words]

    return tokens

tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("AI tokens:")
print(tokens1)

print("\nBlockchain tokens:")
print(tokens2)
```

```python
# Convert tokens into sets
set1 = set(tokens1)
set2 = set(tokens2)

# Find intersection and union
intersection = set1.intersection(set2)
union = set1.union(set2)

# Calculate Jaccard similarity
jaccard = len(intersection) / len(union)

print("Common words:", intersection)
print("Total unique words:", len(union))
print("Jaccard Similarity:", round(jaccard, 4))
```

```python
# Combine both texts
texts = [text1, text2]

# Generate TF-IDF vectors
vectorizer = TfidfVectorizer(
    preprocessor=lambda x: re.sub(r'[^a-z\s]', '', x.lower()),
    tokenizer=word_tokenize,
    token_pattern=None,
    stop_words='english'
)

tfidf_matrix = vectorizer.fit_transform(texts)

# Calculate cosine similarity
cosine_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

print("Cosine Similarity:", round(cosine_sim[0][0], 4))
```

```python
print("Similarity Comparison")
print("---------------------")
print("Jaccard Similarity:", round(jaccard, 4))
print("Cosine Similarity:", round(cosine_sim[0][0], 4))

if jaccard < 0.3 and cosine_sim[0][0] < 0.3:
    print("\nInterpretation: The two texts have low similarity.")
    print("They discuss different technologies with different vocabularies.")
else:
    print("\nInterpretation: The texts share some vocabulary or features.")
```

```python
print("Analysis of Similarity Metrics:\n")

print("Jaccard Similarity:", round(jaccard, 4))
print("Cosine Similarity:", round(cosine_sim[0][0], 4))

print("\nConclusion:")
print("Jaccard similarity compares the common unique words between the two texts.")
print("Cosine similarity compares TF-IDF weighted word vectors.")
print("In this case, cosine similarity provides better insights because")
print("it considers the importance of words in each text.")
print("However, both metrics show lexical similarity rather than full semantic meaning.")
```

```python
from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt
```

```python
reviews = [
    "This smartphone is amazing. The camera is excellent and the battery lasts long.",
    "I love this phone. It is fast, stylish, and very easy to use.",
    "The phone is terrible. The battery drains quickly and the screen is poor.",
    "Very disappointing product. It is slow and the camera quality is bad.",
    "The phone is okay. It has average features and works as expected.",
    "The product is neither good nor bad. It performs ordinary tasks adequately."
]

for i, review in enumerate(reviews, start=1):
    print(f"Review {i}: {review}")
```

```python
polarity_scores = []
subjectivity_scores = []

for review in reviews:
    analysis = TextBlob(review)

    polarity_scores.append(analysis.sentiment.polarity)
    subjectivity_scores.append(analysis.sentiment.subjectivity)

    print("Review:", review)
    print("Polarity:", round(analysis.sentiment.polarity, 3))
    print("Subjectivity:", round(analysis.sentiment.subjectivity, 3))
    print("-" * 60)
```

```python
def classify_sentiment(polarity):
    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"

sentiments = []

for i, review in enumerate(reviews):
    sentiment = classify_sentiment(polarity_scores[i])
    sentiments.append(sentiment)

    print(f"Review {i+1}: {sentiment}")
```

```python
import pandas as pd

results = pd.DataFrame({
    "Review": reviews,
    "Polarity": polarity_scores,
    "Subjectivity": subjectivity_scores,
    "Sentiment": sentiments
})

print(results.to_string(index=False))
```

```python
# Select all positive reviews
positive_reviews = [
    reviews[i]
    for i in range(len(reviews))
    if sentiments[i] == "Positive"
]

# Combine positive reviews into one text
positive_text = " ".join(positive_reviews)

# Generate word cloud
wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white",
    colormap="viridis"
).generate(positive_text)

# Display word cloud
plt.figure(figsize=(12, 6))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of Positive Smartphone Reviews")
plt.show()
```

```python
sentiment_counts = results["Sentiment"].value_counts()

print("Sentiment Summary:")
print(sentiment_counts)

sentiment_counts.plot(kind="bar", title="Review Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.show()
```

```python
import numpy as np
import nltk

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
text = """
Artificial intelligence is changing the world through innovation and discovery.
AI helps computers learn from data and solve complex problems.
Scientists use artificial intelligence to study proteins and understand diseases.
AlphaFold predicts the three dimensional structures of proteins.
This discovery helps researchers understand biology and develop new medicines.
Machine learning allows computers to recognize patterns and make predictions.
Deep learning uses neural networks to learn complex features from information.
Artificial intelligence supports healthcare education science and technology.
Researchers continue to improve intelligent systems for a better future.
Technology and scientific research can improve human life.
"""

print(text)
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

print("Word Index:")
print(tokenizer.word_index)

print("\nTotal unique words:", total_words)

# Convert text into a sequence of integers
token_list = tokenizer.texts_to_sequences([text])[0]

print("\nTokenized Sequence:")
print(token_list)
```

```python
# Split paragraph into sentences
sentences = [s.strip() for s in text.split(".") if s.strip()]

input_sequences = []

for sentence in sentences:
    token_list = tokenizer.texts_to_sequences([sentence])[0]

    # Create n-gram sequences
    for i in range(2, len(token_list) + 1):
        n_gram_sequence = token_list[:i]
        input_sequences.append(n_gram_sequence)

# Pad sequences to equal length
max_sequence_len = max(len(seq) for seq in input_sequences)

input_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding="pre"
)

# Separate predictors and labels
X = input_sequences[:, :-1]
y = input_sequences[:, -1]

print("Input shape:", X.shape)
print("Output shape:", y.shape)
print("Maximum sequence length:", max_sequence_len)
```

```python
model = Sequential([
    Embedding(total_words, 32, input_length=max_sequence_len - 1),
    LSTM(64),
    Dense(total_words, activation="softmax")
])

model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

model.summary()
```

```python
history = model.fit(
    X,
    y,
    epochs=150,
    verbose=1
)
```

```python
def generate_text(seed_text, next_words=10):
    output_text = seed_text

    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences([output_text])[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=max_sequence_len - 1,
            padding="pre",
            truncating="pre"
        )

        predicted_probs = model.predict(token_list, verbose=0)[0]
        predicted_word_index = np.argmax(predicted_probs)

        output_word = None

        for word, index in tokenizer.word_index.items():
            if index == predicted_word_index:
                output_word = word
                break

        if output_word is None:
            break

        output_text += " " + output_word

    return output_text.capitalize() + "."
```

```python
print("Generated Text:\n")

print("1.", generate_text("artificial", 10))
print("2.", generate_text("protein", 10))
print("3.", generate_text("science", 10))
```
