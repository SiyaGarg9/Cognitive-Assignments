```python
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40]) 
corrected_temperature = temperature + 2
print(corrected_temperature)
fahrenheit = (9/5) * corrected_temperature + 32
print(fahrenheit)
above_32 = corrected_temperature[corrected_temperature > 32]
print(above_32)
count = np.sum(corrected_temperature > 32)
print(count)
```

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum:", np.max(steps))
print("Minimum:", np.min(steps))
print("Total for each day:", np.sum(steps, axis=0))
print("Total for each user:", np.sum(steps, axis=1))
print("Maximum index:", np.argmax(steps))
print("Maximum position:", np.unravel_index(np.argmax(steps), steps.shape))
```

```python
import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)

subset = original[1:5]
print("Subset:", subset)

subset[0] = 999
print("After modifying subset:")
print("Original:", original)
print("Subset:", subset)

original = np.array([1, 2, 3, 4, 5, 6])
copied = original[1:5].copy()
copied[0] = 500

print("\nAfter modifying copied array:")
print("Original:", original)
print("Copied:", copied)

matrix = np.arange(1, 13).reshape(3, 4)
print("\n3x4 Matrix:")
print(matrix)

print("\nFirst row:", matrix[0, :])
print("Last row:", matrix[-1, :])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[1:3, 2:4])

flat = matrix.flatten()
rav = matrix.ravel()

print("\nFlatten:", flat)
print("Ravel:", rav)

rav[0] = 100
print("\nAfter modifying ravel:")
print("Ravel:", rav)
print("Matrix:")
print(matrix)

flat[0] = 500
print("\nAfter modifying flatten:")
print("Flatten:", flat)
print("Matrix:")
print(matrix)

print("\nShape:", matrix.shape)
print("ndim:", matrix.ndim)
print("Size:", matrix.size)
print("dtype:", matrix.dtype)
```

```python
import numpy as np
y = np.array([40, 65, 30, 85])

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
print("Shape:", X.shape)
print("Dimensions:", X.ndim)

print("\nX.T:")
print(X.T)

XtX = X.T @ X
print("\nX.T @ X:")
print(XtX)

inverse = np.linalg.inv(XtX)
print("\nInverse:")
print(inverse)

beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("\nBeta coefficients:")
print(beta)

new_user = np.array([5, 40, 7])
prediction = new_user @ beta

print("\nPredicted assistance score:", prediction)
```
