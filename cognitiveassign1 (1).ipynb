```python
def AddOdd(n):
    s=0
    for i in range(1,n+1):
        if i%2!=0:
            s=s+i
    return s

n=int(input("Enter the value of n: "))
print(AddOdd(n))
```

```python
def AddPrime(n):
    s=0
    for i in range(2,n+1):
        prime=True
        for j in range(2,i):
            if i%j==0:
                prime=False
                break
        if prime:
            s=s+i
    return s

n=int(input("Enter the value of n: "))
print(AddPrime(n))
```
